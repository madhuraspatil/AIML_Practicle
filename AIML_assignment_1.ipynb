4. Theory

A Simple Reflex Agent selects an action based only on the current percept (current room condition) without considering past states. In this experiment, the environment consists of two floors:

Floor 1: A1, A2
Floor 2: B1, B2

Each room can be:

0 → Clean
1 → Dirty

The vacuum cleaner starts from a random room, checks the room condition, cleans it if dirty, and then moves to other rooms. Battery energy is consumed during movement and cleaning. Extra energy is consumed when the vacuum changes floors.

The floor is identified from the room name:

Rooms starting with A belong to Floor 1.
Rooms starting with B belong to Floor 2.

This system demonstrates a practical application of intelligent agents in a multi-room environment.

5. Mathematical Formula / Model
Performance Measure

Let:

C = number of rooms cleaned
M = number of movements within the same floor
F = number of floor changes

The performance score is:

P = C - M - 2F

Battery Model

Battery decreases by:

1 unit for cleaning a room
1 unit for movement within the same floor
2 units for changing floors

```python
import random

# Environment with 2 floors and 2 rooms on each floor
class Environment:
    def __init__(self):
        # 0 = Clean, 1 = Dirty
        self.rooms = {
            'A1': random.randint(0, 1),
            'A2': random.randint(0, 1),
            'B1': random.randint(0, 1),
            'B2': random.randint(0, 1)
        }

# Smart Vacuum Cleaner Agent
class SmartVacuumAgent:
    def __init__(self, env):

        print("Initial Environment:", env.rooms)

        battery = 12
        score = 0
        dust_bin = 0

        rooms = list(env.rooms.keys())

        # Start from a random room
        current_room = random.choice(rooms)

        print(f"Vacuum starts at {current_room}")
        print(f"Initial Battery: {battery}\n")

        # Visit all rooms
        for room in rooms:

            # Show current floor
            floor = room[0]

            if floor == 'A':
                print("Currently working on Floor 1")
            else:
                print("Currently working on Floor 2")

            # Move if needed
            if room != current_room:

                current_floor = current_room[0]
                next_floor = room[0]

                if current_floor != next_floor:
                    print(f"Changing floor from Floor {1 if current_floor=='A' else 2} to Floor {1 if next_floor=='A' else 2}")
                    battery -= 2      # extra battery for floor change
                    score -= 2
                else:
                    print(f"Moving within Floor {1 if current_floor=='A' else 2}")
                    battery -= 1
                    score -= 1

                if battery <= 0:
                    print("Battery exhausted! Vacuum stopped.")
                    break

                print(f"Moved to {room}")
                print(f"Battery after moving: {battery}")

                current_room = room

            # Cleaning
            if env.rooms[room] == 1:

                if battery <= 0:
                    print("Battery exhausted! Cannot clean.")
                    break

                print(f"{room} is Dirty")

                env.rooms[room] = 0
                battery -= 1
                score += 1
                dust_bin += 1

                print(f"{room} has been Cleaned")
                print(f"Battery after cleaning: {battery}\n")

            else:
                print(f"{room} is already Clean\n")

        # Final status
        print("Final Environment:", env.rooms)
        print("Dust Collected:", dust_bin)
        print("Battery Remaining:", battery)
        print("Performance Score:", score)

# Run program
env = Environment()
agent = SmartVacuumAgent(env)
```
